```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin
```

Storing url and checking response

```python
url = "https://quotes.toscrape.com/" 
url
```

```python
response = requests.get(url, timeout=30)
response
```

Checking website html code

```python
html = response.text
```

```python
print(html[10000:15000])
```

Turning html into beautifulsoup object 

```python
soup = BeautifulSoup(html, "html.parser")
type(soup)
```

Findind all quotes data

```python
quotes = soup.select("div.quote")
len(quotes)
```

for looping through the quotes

```python
for quote in quotes:
    name = quote.select_one("span small").text
    print(name)
```

Creating rows, then dataframe

```python
rows = []
#for looping quotes to store text, name and tags
for quote in quotes:
    text = quote.select_one("span.text").get_text(strip=True)
    author = quote.select_one("small.author").get_text(strip=True)
    quote_tags = quote.select("a")
    rows.append({
        "quote_text": text,
        "author": author,
        "quote_tags": quote_tags
    })
len(rows)
```

```python
df = pd.DataFrame(rows)
df.head()
```
