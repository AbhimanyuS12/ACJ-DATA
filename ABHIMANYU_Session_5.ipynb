## 1. Why functions?

Imagine you repeatedly calculate percentage change in a reporting project.

Without a function, you might write:

```python
(new_value - old_value) / old_value * 100
```

again and again.

That works—but repeated code creates problems:

- it is easy to make inconsistent changes
- it is harder to debug
- it makes notebooks longer
- if the calculation changes, you have to fix it everywhere

A function lets us write the logic once and reuse it.


### What we are about to do

Lets begin with the simplest possible function.

The keyword `def` means **define a function**.

The function below is called `say_hello`. The indented line is the instruction Python should perform whenever we call that function.

Notice the **colon** after the function name and the **indentation** underneath it—both are required.

````python
def say_hello():
    print("Hello, newsroom!")
````

### What happened?

Nothing printed yet.

That is because **defining** a function and **running** a function are two different things.

The previous cell told Python:

> “Here is a set of instructions called `say_hello`.”

The next cell actually **calls** the function by writing its name followed by parentheses.


````python
say_hello()
````

````python
def onam_snack():
    print("Banana Chips")
````

````python
onam_snack()
````

## 2. Functions become useful when they accept input

Our first function always prints exactly the same thing.

Usually we want a function to behave differently depending on the information we give it.

A **parameter** is a placeholder inside the function definition.

Here, `name` is a parameter. Whatever value we provide when calling the function will temporarily be stored under that name.

````python
def greet(name):
    print("Salaam", name)
````

### Now call the same function with different arguments

An **argument** is the actual value we pass into a function.

In:

```python
greet("Areena")
```

- `name` is the parameter
- `"Areena"` is the argument

The same function can be reused with any name.


````python
greet("Bombay")
````

## 3. A reporting example: percentage change

Suppose revenue rises from ₹100 crore to ₹125 crore.

Percentage change is:

```text
(new - old) / old × 100
```

We could write this calculation every time, but this is exactly the kind of repeated logic that belongs inside a function.

Our function will accept two parameters:

- `old`
- `new`

and calculate the percentage change between them.


````python
def percent_change(old, new):
    change = ((new-old)/old)*100
    return change
````

````python
percent_change(800, 510)
````

### What does `return` mean?

`return` sends a value **back out of the function**.

That is different from `print()`.

- `print()` displays something for us to read.
- `return` gives the result back to Python so we can store it, analyse it, or use it in another calculation.

Let's call the function.


````python
percent_change(55, 65.18)
````

### Store the returned result

Because the function **returns** a value, we can save that value in a variable.

This is one reason `return` is so useful: the output does not simply appear on screen and disappear. It can become input for the next step of our analysis.


## 4. `print()` versus `return`

This distinction is easy to miss at first.

The first function below only **prints** the answer.

The second function **returns** the answer.

Both may look similar when you run them, but only the returned value can easily be reused by Python.


### First: a function that only prints

This function calculates the result and displays it.

We will assign the function call to a variable called `result`. Then we will inspect that variable to see what Python actually stored.


````python
def percent_change(old, new):
    change = ((new-old)/old)*100
    print(change)
````

````python
percent_change(120,240)
````

### Now compare it with a function that returns

Our original `percent_change()` function sends the value back to Python.

That means the variable receives the number itself rather than `None`.


````python
def percent_change(old, new):
    change = ((new-old)/old)*100
    return(change)
percent_change(120,240)
````

## 5. Functions can contain decisions

Functions are not limited to calculations.

They can also use conditions.

Suppose we want a quick label for a percentage change:

- positive → `"increase"`
- negative → `"decrease"`
- zero → `"no change"`

We can put an `if` statement inside a function.


````python
def shift(old, new):
    change = percent_change(old, new)

    if change>0:
        return "up"
    elif change<0:
        return "down"
    else:
        return "same"
````

### Try the function with different values

The same function now follows different paths depending on the arguments we provide.


````python
shift(12,13)
````

````python
print(shift(12,13))
````

````python
shift(12,11)
````

````python
shift(12,12)
````

## 6. Functions can help us clean data

Think back to our Beautiful Soup exercise.

The scraped prices looked like:

```text
£51.77
```

or, because of an encoding issue:

```text
Â£51.77
```

We can create a reusable function that takes a messy price string and returns only the numeric value.

This is useful because data cleaning often involves applying the same rule to many values.


````python
def clean_price(price_text):
    cleaned = {
        price_text
        .replace("Â", "")
        .replace("£", "")
        .strip()
    }

    return float(cleaned)
````

### Test a function before using it on an entire dataset

A good habit is to try the function on one or two known examples first.

If these produce the expected numeric values, we have some confidence that the cleaning logic works before applying it to hundreds or thousands of rows.


````python
clean_price("Â£51.77")
````

# Final recap

A function has four important pieces:

```python
def function_name(parameter):
    # instructions
    return result
```

### Vocabulary

**Function**  
A named, reusable set of instructions.

**`def`**  
Tells Python that we are defining a function.

**Parameter**  
A placeholder in the function definition.

**Argument**  
The actual value supplied when the function is called.

**`return`**  
Sends a result back out of the function.

**Call**  
Run the function by writing its name followed by parentheses.
