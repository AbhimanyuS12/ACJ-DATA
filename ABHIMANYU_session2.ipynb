## 7. Filter rows

## 10. Check missing values

A missing value does not necessarily mean an error. It may mean an indicator was not published for that state or period.

```python
df.isna().sum().sort_values(ascending=False)
```

```python
df[df["life_expectancy_years_total"].isna()][[
    "state",
    "life_expectancy_years_total"
]]
```

```python
df[df["infant_mortality_rate_per_1000"].isna()][[
    "state",
    "infant_mortality_rate_per_1000"
]]
```

## 11. Compare two variables

For now, we are looking for patterns, not claiming causation.

```python
df[[
    "state",
    "multidimensional_poverty_headcount_percent",
    "infant_mortality_rate_per_1000"
]].sort_values(
    "multidimensional_poverty_headcount_percent",
    ascending=False
).head(10)
```

```python
df[[
    "state",
    "infant_mortality_rate_per_1000",
    "natural_population_growth_rate"
]].sort_values(
    "natural_population_growth_rate",
    ascending=True
).head(5)
```
